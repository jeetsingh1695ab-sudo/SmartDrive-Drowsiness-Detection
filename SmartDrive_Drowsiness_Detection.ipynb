```python
# ============================================
# SMARTDRIVE - STAGE 3.1
# Check Open / Closed Eye Labels
# ============================================

import os
from collections import Counter

extract_path = "/content/mrl_dataset"

image_extensions = (".png", ".jpg", ".jpeg", ".bmp")

labels = []

for root, dirs, files in os.walk(extract_path):
    for file in files:
        if file.lower().endswith(image_extensions):

            # MRL filename format:
            # ...-011-0-...  → 0 = closed
            # ...-011-1-...  → 1 = open

            parts = file.split("-")

            if len(parts) >= 3:
                eye_state = parts[2]

                if eye_state in ["0", "1"]:
                    labels.append(int(eye_state))

label_counts = Counter(labels)

print("👁️ SmartDrive - Eye Label Verification")
print("=" * 50)

print("🔴 Closed eyes (0):", label_counts[0])
print("🟢 Open eyes (1):", label_counts[1])
print("📊 Total labelled images:", len(labels))

if 0 in label_counts and 1 in label_counts:
    print("\n✅ BOTH OPEN AND CLOSED EYE LABELS FOUND!")
    print("🎉 Dataset is correctly labelled.")
else:
    print("\n❌ Label verification failed.")
```

```python
# ============================================
# SMARTDRIVE - STAGE 3.1 (CORRECTED)
# Verify Open / Closed Eye Labels
# ============================================

import os
from collections import Counter

extract_path = "/content/mrl_dataset"

labels = []

for root, dirs, files in os.walk(extract_path):
    for file in files:
        if file.lower().endswith(".png"):

            # Example:
            # s0002_00001_0_0_0_0_0_01.png
            parts = file.replace(".png", "").split("_")

            # Eye state = 5th attribute
            # Python index 4
            if len(parts) >= 8:
                eye_state = parts[4]

                if eye_state in ["0", "1"]:
                    labels.append(int(eye_state))

label_counts = Counter(labels)

print("👁️ SmartDrive - Eye Label Verification")
print("=" * 50)

print("🔴 Closed eyes (0):", label_counts[0])
print("🟢 Open eyes (1):", label_counts[1])
print("📊 Total labelled images:", len(labels))

if 0 in label_counts and 1 in label_counts:
    print("\n✅ BOTH OPEN AND CLOSED EYE LABELS FOUND!")
    print("🎉 Dataset is correctly labelled.")
else:
    print("\n❌ Label verification failed.")
```

```python
# ============================================
# SMARTDRIVE - STAGE 3.4
# Create Image Paths & Labels
# ============================================

import os
import pandas as pd

extract_path = "/content/mrl_dataset"

image_paths = []
labels = []

for root, dirs, files in os.walk(extract_path):

    for file in files:

        if file.lower().endswith(".png"):

            # Remove extension and split filename
            filename = os.path.splitext(file)[0]
            parts = filename.split("_")

            # Eye state = 5th attribute
            # 0 = Closed
            # 1 = Open
            if len(parts) >= 8:

                eye_state = parts[4]

                if eye_state in ["0", "1"]:

                    image_paths.append(os.path.join(root, file))
                    labels.append(int(eye_state))

# Create DataFrame
data = pd.DataFrame({
    "image_path": image_paths,
    "label": labels
})

print("👁️ SmartDrive - Dataset Preparation")
print("=" * 50)

print("📊 Total images:", len(data))
print("🔴 Closed eyes (0):", (data["label"] == 0).sum())
print("🟢 Open eyes (1):", (data["label"] == 1).sum())

print("\n📋 Dataset preview:")
print(data.head())

print("\n✅ Image paths and labels successfully organized!")
```

```python
# ============================================
# SMARTDRIVE - STAGE 4.1
# Train / Validation / Test Split
# ============================================

from sklearn.model_selection import train_test_split

# First split: 80% Training, 20% Temporary
train_data, temp_data = train_test_split(
    data,
    test_size=0.20,
    random_state=42,
    stratify=data["label"]
)

# Second split: Temporary → 10% Validation, 10% Test
val_data, test_data = train_test_split(
    temp_data,
    test_size=0.50,
    random_state=42,
    stratify=temp_data["label"]
)

print("🚗 SmartDrive - Dataset Split")
print("=" * 50)

print("📚 Training images:", len(train_data))
print("🔍 Validation images:", len(val_data))
print("🧪 Test images:", len(test_data))

print("\n📊 Total images:", len(train_data) + len(val_data) + len(test_data))

print("\n✅ Dataset successfully split into Train / Validation / Test!")
```

```python
# ============================================
# SMARTDRIVE - STAGE 5.1
# CNN Data Generators - CORRECTED
# ============================================

import tensorflow as tf
from tensorflow.keras.preprocessing.image import ImageDataGenerator

IMG_SIZE = (64, 64)
BATCH_SIZE = 32

# Convert labels to strings for Keras
train_data = train_data.copy()
val_data = val_data.copy()
test_data = test_data.copy()

train_data["label"] = train_data["label"].astype(str)
val_data["label"] = val_data["label"].astype(str)
test_data["label"] = test_data["label"].astype(str)

# Data generators
train_datagen = ImageDataGenerator(
    rescale=1.0 / 255.0
)

val_datagen = ImageDataGenerator(
    rescale=1.0 / 255.0
)

test_datagen = ImageDataGenerator(
    rescale=1.0 / 255.0
)

# Training generator
train_generator = train_datagen.flow_from_dataframe(
    dataframe=train_data,
    x_col="image_path",
    y_col="label",
    target_size=IMG_SIZE,
    batch_size=BATCH_SIZE,
    class_mode="binary",
    color_mode="grayscale",
    shuffle=True,
    seed=42
)

# Validation generator
val_generator = val_datagen.flow_from_dataframe(
    dataframe=val_data,
    x_col="image_path",
    y_col="label",
    target_size=IMG_SIZE,
    batch_size=BATCH_SIZE,
    class_mode="binary",
    color_mode="grayscale",
    shuffle=False
)

# Test generator
test_generator = test_datagen.flow_from_dataframe(
    dataframe=test_data,
    x_col="image_path",
    y_col="label",
    target_size=IMG_SIZE,
    batch_size=BATCH_SIZE,
    class_mode="binary",
    color_mode="grayscale",
    shuffle=False
)

print("\n✅ CNN data generators created successfully!")
print("📐 Image size:", IMG_SIZE)
print("📦 Batch size:", BATCH_SIZE)
print("🏷️ Class mapping:", train_generator.class_indices)
```

```python
# ============================================
# SMARTDRIVE - STAGE 5.2
# CNN Model Architecture
# ============================================

from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import (
    Conv2D,
    MaxPooling2D,
    Flatten,
    Dense,
    Dropout
)

# Create CNN model
model = Sequential([

    # First Convolution Block
    Conv2D(32, (3, 3), activation="relu", input_shape=(64, 64, 1)),
    MaxPooling2D(pool_size=(2, 2)),

    # Second Convolution Block
    Conv2D(64, (3, 3), activation="relu"),
    MaxPooling2D(pool_size=(2, 2)),

    # Third Convolution Block
    Conv2D(128, (3, 3), activation="relu"),
    MaxPooling2D(pool_size=(2, 2)),

    # Convert feature maps to 1D
    Flatten(),

    # Fully Connected Layer
    Dense(128, activation="relu"),
    Dropout(0.5),

    # Output Layer
    Dense(1, activation="sigmoid")
])

# Display model architecture
model.summary()

print("\n✅ CNN model created successfully!")
```

```python
# ============================================
# SMARTDRIVE - STAGE 5.3
# Compile CNN Model
# ============================================

model.compile(
    optimizer="adam",
    loss="binary_crossentropy",
    metrics=["accuracy"]
)

print("🚗 SmartDrive")
print("Stage 5.3: CNN Model Compilation")
print("=" * 45)

print("✅ Optimizer: Adam")
print("✅ Loss Function: Binary Crossentropy")
print("✅ Metric: Accuracy")
print("\n🎉 CNN model compiled successfully!")
```

```python
# ============================================
# SMARTDRIVE - STAGE 5.2 + 5.3
# Create AND Compile CNN Model
# ============================================

from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Conv2D, MaxPooling2D, Flatten, Dense, Dropout

# Create CNN model
model = Sequential([
    Conv2D(32, (3, 3), activation="relu", input_shape=(64, 64, 1)),
    MaxPooling2D(pool_size=(2, 2)),

    Conv2D(64, (3, 3), activation="relu"),
    MaxPooling2D(pool_size=(2, 2)),

    Conv2D(128, (3, 3), activation="relu"),
    MaxPooling2D(pool_size=(2, 2)),

    Flatten(),

    Dense(128, activation="relu"),
    Dropout(0.5),

    Dense(1, activation="sigmoid")
])

# Compile CNN model
model.compile(
    optimizer="adam",
    loss="binary_crossentropy",
    metrics=["accuracy"]
)

print("🚗 SmartDrive")
print("=" * 50)
print("✅ CNN model created")
print("✅ CNN model compiled")
print("✅ Optimizer: Adam")
print("✅ Loss: Binary Crossentropy")
print("✅ Metric: Accuracy")
print("\n🎉 MODEL IS READY FOR TRAINING!")
```

```python
# ============================================
# SMARTDRIVE - STAGE 5.1
# Create Data Generators
# ============================================

import os
import pandas as pd
from sklearn.model_selection import train_test_split
from tensorflow.keras.preprocessing.image import ImageDataGenerator

extract_path = "/content/mrl_dataset"

# Create image paths and labels
image_paths = []
labels = []

for root, dirs, files in os.walk(extract_path):
    for file in files:
        if file.lower().endswith(".png"):

            filename = os.path.splitext(file)[0]
            parts = filename.split("_")

            if len(parts) >= 8:
                eye_state = parts[4]

                if eye_state in ["0", "1"]:
                    image_paths.append(os.path.join(root, file))
                    labels.append(int(eye_state))

data = pd.DataFrame({
    "image_path": image_paths,
    "label": labels
})

# Train / Validation / Test split
train_data, temp_data = train_test_split(
    data,
    test_size=0.20,
    random_state=42,
    stratify=data["label"]
)

val_data, test_data = train_test_split(
    temp_data,
    test_size=0.50,
    random_state=42,
    stratify=temp_data["label"]
)

# Convert labels to strings for Keras
train_data["label"] = train_data["label"].astype(str)
val_data["label"] = val_data["label"].astype(str)
test_data["label"] = test_data["label"].astype(str)

# Image settings
IMG_SIZE = (64, 64)
BATCH_SIZE = 32

# Data generators
train_datagen = ImageDataGenerator(rescale=1.0 / 255.0)
val_datagen = ImageDataGenerator(rescale=1.0 / 255.0)
test_datagen = ImageDataGenerator(rescale=1.0 / 255.0)

train_generator = train_datagen.flow_from_dataframe(
    train_data,
    x_col="image_path",
    y_col="label",
    target_size=IMG_SIZE,
    batch_size=BATCH_SIZE,
    class_mode="binary",
    color_mode="grayscale",
    shuffle=True,
    seed=42
)

val_generator = val_datagen.flow_from_dataframe(
    val_data,
    x_col="image_path",
    y_col="label",
    target_size=IMG_SIZE,
    batch_size=BATCH_SIZE,
    class_mode="binary",
    color_mode="grayscale",
    shuffle=False
)

test_generator = test_datagen.flow_from_dataframe(
    test_data,
    x_col="image_path",
    y_col="label",
    target_size=IMG_SIZE,
    batch_size=BATCH_SIZE,
    class_mode="binary",
    color_mode="grayscale",
    shuffle=False
)

print("\n🎉 DATA GENERATORS READY!")
print("✅ Train generator ready")
print("✅ Validation generator ready")
print("✅ Test generator ready")
```

```python
# ============================================
# SMARTDRIVE - STAGE 6.1
# CNN Model Training
# ============================================

print("🚗 SmartDrive")
print("Stage 6.1: CNN Model Training")
print("=" * 50)

EPOCHS = 10

history = model.fit(
    train_generator,
    validation_data=val_generator,
    epochs=EPOCHS,
    verbose=1
)

print("\n🎉 CNN training completed successfully!")
```

```python
# ============================================
# SMARTDRIVE - STAGE 6.2
# Training Performance Graphs
# ============================================

import matplotlib.pyplot as plt

# Training history
train_accuracy = history.history["accuracy"]
val_accuracy = history.history["val_accuracy"]

train_loss = history.history["loss"]
val_loss = history.history["val_loss"]

epochs = range(1, len(train_accuracy) + 1)

# --------------------------------------------
# Accuracy Graph
# --------------------------------------------

plt.figure(figsize=(10, 5))

plt.plot(epochs, train_accuracy, label="Training Accuracy")
plt.plot(epochs, val_accuracy, label="Validation Accuracy")

plt.title("SmartDrive - Training vs Validation Accuracy")
plt.xlabel("Epoch")
plt.ylabel("Accuracy")
plt.legend()
plt.grid(True)

plt.show()

# --------------------------------------------
# Loss Graph
# --------------------------------------------

plt.figure(figsize=(10, 5))

plt.plot(epochs, train_loss, label="Training Loss")
plt.plot(epochs, val_loss, label="Validation Loss")

plt.title("SmartDrive - Training vs Validation Loss")
plt.xlabel("Epoch")
plt.ylabel("Loss")
plt.legend()
plt.grid(True)

plt.show()

print("✅ Training performance graphs generated successfully!")
```

```python
# ============================================
# SMARTDRIVE - STAGE 7.1
# Test Dataset Evaluation
# ============================================

print("🚗 SmartDrive")
print("Stage 7.1: Model Evaluation")
print("=" * 50)

# Evaluate model on unseen test data
test_loss, test_accuracy = model.evaluate(
    test_generator,
    verbose=1
)

print("\n" + "=" * 50)
print("📊 FINAL TEST RESULTS")
print("=" * 50)

print(f"📉 Test Loss     : {test_loss:.4f}")
print(f"🎯 Test Accuracy : {test_accuracy * 100:.2f}%")

print("\n✅ Model evaluation completed successfully!")
```

```python
# ============================================
# SMARTDRIVE - STAGE 9.2
# CNN Eye State Prediction
# ============================================

import os
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

from sklearn.model_selection import train_test_split
from tensorflow.keras.utils import load_img, img_to_array

print("🚗 SmartDrive")
print("Stage 9.2: CNN Eye State Prediction")
print("=" * 50)

# --------------------------------------------
# STEP 1: Check dataset
# --------------------------------------------

dataset_path = "/content/mrl_dataset"

if not os.path.exists(dataset_path):
    print("❌ Dataset folder not found!")
else:
    print("✅ Dataset found!")

# --------------------------------------------
# STEP 2: Create image dataframe
# --------------------------------------------

image_paths = []
labels = []

for root, dirs, files in os.walk(dataset_path):

    for file in files:

        if file.lower().endswith(".png"):

            filename = os.path.splitext(file)[0]
            parts = filename.split("_")

            if len(parts) >= 8:

                eye_state = parts[4]

                if eye_state in ["0", "1"]:

                    image_paths.append(
                        os.path.join(root, file)
                    )

                    labels.append(
                        int(eye_state)
                    )

data = pd.DataFrame({
    "image_path": image_paths,
    "label": labels
})

print("📊 Total images:", len(data))

# --------------------------------------------
# STEP 3: Create test dataset
# --------------------------------------------

_, temp_data = train_test_split(
    data,
    test_size=0.20,
    random_state=42,
    stratify=data["label"]
)

_, test_data = train_test_split(
    temp_data,
    test_size=0.50,
    random_state=42,
    stratify=temp_data["label"]
)

print("🔵 Test images:", len(test_data))

# --------------------------------------------
# STEP 4: Check trained model
# --------------------------------------------

if "model" not in globals():

    model_path = "/content/models/smartdrive_eye_model.keras"

    if os.path.exists(model_path):

        from tensorflow.keras.models import load_model

        model = load_model(model_path)

        print("✅ Trained CNN model loaded!")

    else:

        print("❌ Trained model file not found.")
        print("Expected:", model_path)

# --------------------------------------------
# STEP 5: Select test image
# --------------------------------------------

test_image_path = test_data.iloc[0]["image_path"]
actual_label = int(test_data.iloc[0]["label"])

print("\n📷 Testing image:")
print(test_image_path)

# --------------------------------------------
# STEP 6: Load image
# --------------------------------------------

img = load_img(
    test_image_path,
    target_size=(64, 64),
    color_mode="grayscale"
)

img_array = img_to_array(img)

# Normalize
img_array = img_array / 255.0

# Add batch dimension
img_input = np.expand_dims(
    img_array,
    axis=0
)

# --------------------------------------------
# STEP 7: CNN prediction
# --------------------------------------------

prediction = model.predict(
    img_input,
    verbose=0
)[0][0]

predicted_label = 1 if prediction >= 0.5 else 0

# --------------------------------------------
# STEP 8: Convert labels
# --------------------------------------------

actual_name = (
    "OPEN" if actual_label == 1
    else "CLOSED"
)

predicted_name = (
    "OPEN" if predicted_label == 1
    else "CLOSED"
)

confidence = (
    prediction
    if predicted_label == 1
    else 1 - prediction
)

# --------------------------------------------
# STEP 9: Show result
# --------------------------------------------

print("\n" + "=" * 50)
print("📊 CNN EYE PREDICTION RESULT")
print("=" * 50)

print("Actual Eye State    :", actual_name)
print("Predicted Eye State :", predicted_name)
print(f"Confidence          : {confidence * 100:.2f}%")

# --------------------------------------------
# STEP 10: Display image
# --------------------------------------------

plt.figure(figsize=(5, 5))

plt.imshow(
    img_array.squeeze(),
    cmap="gray"
)

plt.axis("off")

plt.title(
    f"Predicted: {predicted_name}\n"
    f"Confidence: {confidence * 100:.2f}%"
)

plt.show()

print("\n✅ Stage 9.2 completed successfully!")
```

```python
# ============================================
# SMARTDRIVE - STAGE 10.1
# COMPLETE WEBCAM + CNN PREDICTION
# ============================================

from IPython.display import display, Javascript
from google.colab.output import eval_js
from base64 import b64decode

import cv2
import numpy as np
import matplotlib.pyplot as plt


# --------------------------------------------
# 1. Webcam function
# --------------------------------------------

def take_photo(quality=0.8):

    js = Javascript("""
    async function takePhoto(quality) {

        const video = document.createElement('video');
        video.style.display = 'block';

        const stream = await navigator.mediaDevices.getUserMedia({
            video: true
        });

        document.body.appendChild(video);

        video.srcObject = stream;

        await video.play();

        await new Promise(
            resolve => setTimeout(resolve, 1000)
        );

        const canvas = document.createElement('canvas');

        canvas.width = video.videoWidth;
        canvas.height = video.videoHeight;

        canvas
            .getContext('2d')
            .drawImage(
                video,
                0,
                0,
                canvas.width,
                canvas.height
            );

        stream
            .getVideoTracks()[0]
            .stop();

        video.remove();

        return canvas.toDataURL(
            'image/jpeg',
            quality
        );
    }
    """)

    display(js)

    data = eval_js(
        'takePhoto({})'.format(quality)
    )

    binary = b64decode(
        data.split(',')[1]
    )

    image = np.frombuffer(
        binary,
        dtype=np.uint8
    )

    frame = cv2.imdecode(
        image,
        cv2.IMREAD_COLOR
    )

    return frame


# --------------------------------------------
# 2. Capture webcam image
# --------------------------------------------

print("🚗 SmartDrive")
print("Stage 10.1: Webcam + CNN Prediction")
print("=" * 50)

print("📷 Opening webcam...")

frame = take_photo()

print("✅ Webcam frame captured!")
print("Frame size:", frame.shape)


# --------------------------------------------
# 3. Convert to grayscale
# --------------------------------------------

gray = cv2.cvtColor(
    frame,
    cv2.COLOR_BGR2GRAY
)


# --------------------------------------------
# 4. Center crop
# --------------------------------------------

height, width = gray.shape

crop_size = min(
    height,
    width
)

start_y = (
    height - crop_size
) // 2

start_x = (
    width - crop_size
) // 2

eye_crop = gray[
    start_y:start_y + crop_size,
    start_x:start_x + crop_size
]


# --------------------------------------------
# 5. Resize to CNN input
# --------------------------------------------

eye_image = cv2.resize(
    eye_crop,
    (64, 64)
)


# --------------------------------------------
# 6. Normalize
# --------------------------------------------

eye_input = (
    eye_image.astype("float32")
    / 255.0
)


# Add channel dimension
eye_input = np.expand_dims(
    eye_input,
    axis=-1
)


# Add batch dimension
eye_input = np.expand_dims(
    eye_input,
    axis=0
)


# --------------------------------------------
# 7. CNN prediction
# --------------------------------------------

prediction = model.predict(
    eye_input,
    verbose=0
)[0][0]


# --------------------------------------------
# 8. Determine eye state
# --------------------------------------------

if prediction >= 0.5:

    eye_state = "OPEN"

    confidence = prediction

else:

    eye_state = "CLOSED"

    confidence = 1.0 - prediction


# --------------------------------------------
# 9. Display result
# --------------------------------------------

print("\n" + "=" * 50)
print("👁️ CNN EYE RESULT")
print("=" * 50)

print("Eye State  :", eye_state)

print(
    f"Confidence : "
    f"{confidence * 100:.2f}%"
)


# --------------------------------------------
# 10. Show image
# --------------------------------------------

plt.figure(
    figsize=(6, 6)
)

plt.imshow(
    eye_image,
    cmap="gray"
)

plt.axis("off")

plt.title(
    f"SmartDrive\n"
    f"Eye: {eye_state}\n"
    f"Confidence: "
    f"{confidence * 100:.2f}%"
)

plt.show()


print("\n✅ Webcam + CNN prediction completed!")
```

```python
# ============================================
# SMARTDRIVE - STAGE 10.3
# REAL-TIME MONITORING + BLINK COUNTER
# ============================================

from IPython.display import display, Javascript
from google.colab.output import eval_js
from base64 import b64decode

import cv2
import numpy as np
import time
import matplotlib.pyplot as plt

# --------------------------------------------
# Webcam capture function
# --------------------------------------------

def take_photo(quality=0.8):
    js = Javascript("""
    async function takePhoto(quality) {
        const video = document.createElement('video');
        video.style.display = 'block';

        const stream = await navigator.mediaDevices.getUserMedia({
            video: true
        });

        document.body.appendChild(video);
        video.srcObject = stream;

        await video.play();
        await new Promise(resolve => setTimeout(resolve, 500));

        const canvas = document.createElement('canvas');
        canvas.width = video.videoWidth;
        canvas.height = video.videoHeight;

        canvas.getContext('2d').drawImage(
            video, 0, 0,
            canvas.width, canvas.height
        );

        stream.getVideoTracks()[0].stop();
        video.remove();

        return canvas.toDataURL('image/jpeg', quality);
    }
    """)

    display(js)

    data = eval_js(
        'takePhoto({})'.format(quality)
    )

    binary = b64decode(data.split(',')[1])

    image = np.frombuffer(
        binary,
        dtype=np.uint8
    )

    frame = cv2.imdecode(
        image,
        cv2.IMREAD_COLOR
    )

    return frame


# --------------------------------------------
# Settings
# --------------------------------------------

TOTAL_FRAMES = 10
CLOSED_TIME_THRESHOLD = 2.0

blink_count = 0
closed_start = None
previous_state = "OPEN"

print("🚗 SmartDrive")
print("Stage 10.3: Real-Time Eye Monitoring")
print("=" * 50)

print("📷 Webcam monitoring starting...")
print("📌 Please look at the camera.")
print()


# --------------------------------------------
# Monitoring loop
# --------------------------------------------

for frame_number in range(TOTAL_FRAMES):

    frame = take_photo()

    if frame is None:
        print("❌ Frame capture failed.")
        continue

    # Convert to grayscale
    gray = cv2.cvtColor(
        frame,
        cv2.COLOR_BGR2GRAY
    )

    # Center crop
    height, width = gray.shape

    crop_size = min(
        height,
        width
    )

    start_y = (height - crop_size) // 2
    start_x = (width - crop_size) // 2

    eye_crop = gray[
        start_y:start_y + crop_size,
        start_x:start_x + crop_size
    ]

    # Resize for CNN
    eye_image = cv2.resize(
        eye_crop,
        (64, 64)
    )

    # Normalize
    eye_input = (
        eye_image.astype("float32") / 255.0
    )

    eye_input = np.expand_dims(
        eye_input,
        axis=-1
    )

    eye_input = np.expand_dims(
        eye_input,
        axis=0
    )

    # CNN prediction
    prediction = model.predict(
        eye_input,
        verbose=0
    )[0][0]

    if prediction >= 0.5:
        current_state = "OPEN"
        confidence = prediction
    else:
        current_state = "CLOSED"
        confidence = 1.0 - prediction

    # ----------------------------------------
    # Blink detection
    # ----------------------------------------

    if (
        previous_state == "OPEN"
        and current_state == "CLOSED"
    ):
        closed_start = time.time()

    if (
        previous_state == "CLOSED"
        and current_state == "OPEN"
    ):
        if closed_start is not None:

            closed_duration = (
                time.time() - closed_start
            )

            if closed_duration < CLOSED_TIME_THRESHOLD:
                blink_count += 1

            closed_start = None

    # ----------------------------------------
    # Drowsiness detection
    # ----------------------------------------

    drowsiness = False

    if current_state == "CLOSED":

        if closed_start is not None:

            closed_duration = (
                time.time() - closed_start
            )

            if closed_duration >= CLOSED_TIME_THRESHOLD:
                drowsiness = True

    # ----------------------------------------
    # Display result
    # ----------------------------------------

    plt.figure(figsize=(5, 5))

    plt.imshow(
        eye_image,
        cmap="gray"
    )

    plt.axis("off")

    if drowsiness:
        title = (
            "🚨 DROWSINESS DETECTED\n"
            f"Blink Count: {blink_count}"
        )
    else:
        title = (
            f"Eye: {current_state}\n"
            f"Confidence: {confidence * 100:.1f}%\n"
            f"Blink Count: {blink_count}"
        )

    plt.title(title)
    plt.show()

    print(
        f"Frame {frame_number + 1}/{TOTAL_FRAMES} | "
        f"Eye: {current_state} | "
        f"Confidence: {confidence * 100:.1f}% | "
        f"Blinks: {blink_count}"
    )

    if drowsiness:
        print("🚨 WARNING: Prolonged eye closure detected!")

    previous_state = current_state

    time.sleep(0.2)


# --------------------------------------------
# Final summary
# --------------------------------------------

print("\n" + "=" * 50)
print("📊 SMARTDRIVE SESSION SUMMARY")
print("=" * 50)

print("👁️ Total Frames Checked :", TOTAL_FRAMES)
print("😉 Blink Count          :", blink_count)

if drowsiness:
    print("🚨 Drowsiness           : DETECTED")
else:
    print("✅ Drowsiness           : NOT DETECTED")

print("=" * 50)
print("✅ Stage 10.3 completed!")
```

```python
# ============================================
# SMARTDRIVE - STAGE 10.10
# ACTUAL FACE + EYE LANDMARK DETECTION
# ============================================

from IPython.display import display, Javascript
from google.colab.output import eval_js
from base64 import b64decode
import cv2
import numpy as np
import matplotlib.pyplot as plt

print("🚗 SmartDrive")
print("Stage 10.10 - Face & Eye Detection")
print("=" * 50)

# ------------------------------------------------
# Camera capture
# ------------------------------------------------

def take_photo(quality=0.8):

    js = Javascript("""
    async function takePhoto(quality) {

        const video = document.createElement('video');

        const stream =
            await navigator.mediaDevices.getUserMedia({
                video: true
            });

        document.body.appendChild(video);

        video.srcObject = stream;

        await video.play();

        await new Promise(
            resolve => setTimeout(resolve, 1000)
        );

        const canvas =
            document.createElement('canvas');

        canvas.width = video.videoWidth;
        canvas.height = video.videoHeight;

        canvas.getContext('2d').drawImage(
            video,
            0,
            0,
            canvas.width,
            canvas.height
        );

        stream.getVideoTracks()[0].stop();
        video.remove();

        return canvas.toDataURL(
            'image/jpeg',
            quality
        );
    }
    """)

    display(js)

    data = eval_js(
        'takePhoto({})'.format(quality)
    )

    binary = b64decode(
        data.split(',')[1]
    )

    image = np.frombuffer(
        binary,
        dtype=np.uint8
    )

    frame = cv2.imdecode(
        image,
        cv2.IMREAD_COLOR
    )

    return frame


# ------------------------------------------------
# MediaPipe New Tasks API
# ------------------------------------------------

from mediapipe.tasks import python
from mediapipe.tasks.python import vision

print("📷 Opening webcam...")

frame = take_photo()

print("✅ Frame captured!")

# ------------------------------------------------
# Display captured image
# ------------------------------------------------

frame_rgb = cv2.cvtColor(
    frame,
    cv2.COLOR_BGR2RGB
)

plt.figure(figsize=(10, 7))

plt.imshow(frame_rgb)

plt.axis("off")

plt.title("SmartDrive - Captured Frame")

plt.show()

print("=" * 50)
print("✅ Webcam capture completed!")
print("🎯 Next step: actual eye landmark detection.")
```

```python
# Stage 10.13 - Drowsiness Alarm Test

from IPython.display import Audio, display
import numpy as np
import wave

def create_beep(filename="/content/drowsiness_alert.wav", duration=0.5, frequency=1000):
    sample_rate = 44100
    t = np.linspace(0, duration, int(sample_rate * duration), endpoint=False)
    audio = (0.5 * np.sin(2 * np.pi * frequency * t) * 32767).astype(np.int16)

    with wave.open(filename, "w") as wav:
        wav.setnchannels(1)
        wav.setsampwidth(2)
        wav.setframerate(sample_rate)
        wav.writeframes(audio.tobytes())

    return filename


# Create alert sound
alert_file = create_beep()

# Current drowsiness state
drowsy = True

if drowsy:
    print("⚠️ DROWSINESS DETECTED!")
    print("🔔 Alert sound playing...")
    display(Audio(alert_file, autoplay=True))
else:
    print("✅ Eyes are normal. No alert.")
```

```python
# SmartDrive — Both Eye Detection Test

import cv2
import numpy as np
import matplotlib.pyplot as plt
from IPython.display import display, Javascript
from google.colab.output import eval_js
from base64 import b64decode

# ---------- CAMERA ----------
def take_photo(quality=0.8):
    js = Javascript("""
    async function takePhoto(quality) {
        const video = document.createElement('video');
        video.style.display = 'block';

        const stream = await navigator.mediaDevices.getUserMedia({
            video: true
        });

        document.body.appendChild(video);
        video.srcObject = stream;

        await video.play();
        await new Promise(resolve => setTimeout(resolve, 800));

        const canvas = document.createElement('canvas');
        canvas.width = video.videoWidth;
        canvas.height = video.videoHeight;

        canvas.getContext('2d').drawImage(
            video, 0, 0, canvas.width, canvas.height
        );

        stream.getVideoTracks()[0].stop();
        video.remove();

        return canvas.toDataURL('image/jpeg', quality);
    }
    """)

    display(js)

    data = eval_js("takePhoto({})".format(quality))
    binary = b64decode(data.split(",")[1])

    image = np.frombuffer(binary, dtype=np.uint8)
    frame = cv2.imdecode(image, cv2.IMREAD_COLOR)

    return frame


# ---------- CAPTURE ----------
print("📸 Camera starting...")
frame = take_photo()

if frame is None:
    raise RuntimeError("❌ Camera frame capture failed.")

print("✅ Frame captured!")


# ---------- FACE LANDMARK DETECTION ----------
rgb = cv2.cvtColor(frame, cv2.COLOR_BGR2RGB)

mp_image = __import__("mediapipe").Image(
    image_format=__import__("mediapipe").ImageFormat.SRGB,
    data=rgb
)

result = face_landmarker.detect(mp_image)

print("Faces detected:", len(result.face_landmarks))


# ---------- EYE CROPPING ----------
if len(result.face_landmarks) == 0:

    print("❌ Face not detected.")
    print("💡 Face camera ke saamne rakho aur lighting achhi rakho.")

else:

    print("✅ FACE DETECTED!")

    landmarks = result.face_landmarks[0]
    h, w = frame.shape[:2]

    left_ids = [33, 133, 159, 145]
    right_ids = [362, 263, 386, 374]

    def get_eye(indices):
        xs = [int(landmarks[i].x * w) for i in indices]
        ys = [int(landmarks[i].y * h) for i in indices]

        x1 = max(min(xs) - 25, 0)
        x2 = min(max(xs) + 25, w)
        y1 = max(min(ys) - 25, 0)
        y2 = min(max(ys) + 25, h)

        return frame[y1:y2, x1:x2]

    left_eye = get_eye(left_ids)
    right_eye = get_eye(right_ids)

    if left_eye.size == 0 or right_eye.size == 0:
        print("❌ Eye crop failed.")
    else:
        print("👁️ LEFT EYE detected!")
        print("👁️ RIGHT EYE detected!")

        plt.figure(figsize=(8, 4))

        plt.subplot(1, 2, 1)
        plt.imshow(cv2.cvtColor(left_eye, cv2.COLOR_BGR2RGB))
        plt.title("LEFT EYE")
        plt.axis("off")

        plt.subplot(1, 2, 2)
        plt.imshow(cv2.cvtColor(right_eye, cv2.COLOR_BGR2RGB))
        plt.title("RIGHT EYE")
        plt.axis("off")

        plt.show()

        print("✅ BOTH EYES DETECTED SUCCESSFULLY!")
```

```python
# SmartDrive — Prepare Train / Validation / Test Data

import os
import pandas as pd
from sklearn.model_selection import train_test_split
from tensorflow.keras.preprocessing.image import ImageDataGenerator

DATASET_PATH = "/content/mrl_dataset"

image_paths = []
labels = []

# Find all MRL images
for root, dirs, files in os.walk(DATASET_PATH):
    for file in files:
        if file.lower().endswith(".png"):
            parts = os.path.splitext(file)[0].split("_")

            # MRL filename: ..._0_... = closed, ..._1_... = open
            if len(parts) > 4 and parts[4] in ["0", "1"]:
                image_paths.append(os.path.join(root, file))
                labels.append(parts[4])

data = pd.DataFrame({
    "image_path": image_paths,
    "label": labels
})

print("🖼️ Total images:", len(data))

if len(data) == 0:
    raise ValueError("❌ Images nahi mili. Dataset extraction check karo.")

# Split: 80% train, 10% validation, 10% test
train_data, temp_data = train_test_split(
    data,
    test_size=0.20,
    random_state=42,
    stratify=data["label"]
)

val_data, test_data = train_test_split(
    temp_data,
    test_size=0.50,
    random_state=42,
    stratify=temp_data["label"]
)

print("Train:", len(train_data))
print("Validation:", len(val_data))
print("Test:", len(test_data))

# Image generators
IMG_SIZE = (64, 64)
BATCH_SIZE = 32

train_datagen = ImageDataGenerator(rescale=1.0 / 255.0)
test_datagen = ImageDataGenerator(rescale=1.0 / 255.0)

train_generator = train_datagen.flow_from_dataframe(
    train_data,
    x_col="image_path",
    y_col="label",
    target_size=IMG_SIZE,
    color_mode="grayscale",
    class_mode="binary",
    batch_size=BATCH_SIZE,
    shuffle=True,
    seed=42
)

val_generator = test_datagen.flow_from_dataframe(
    val_data,
    x_col="image_path",
    y_col="label",
    target_size=IMG_SIZE,
    color_mode="grayscale",
    class_mode="binary",
    batch_size=BATCH_SIZE,
    shuffle=False
)

test_generator = test_datagen.flow_from_dataframe(
    test_data,
    x_col="image_path",
    y_col="label",
    target_size=IMG_SIZE,
    color_mode="grayscale",
    class_mode="binary",
    batch_size=BATCH_SIZE,
    shuffle=False
)

print("\n✅ DATA GENERATORS READY!")
print("Class mapping:", train_generator.class_indices)
```

```python
# SmartDrive — Train CNN + Save Model

import os
import tensorflow as tf
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Conv2D, MaxPooling2D, Flatten, Dense, Dropout

# ---------- CNN ----------
model = Sequential([
    Conv2D(32, (3, 3), activation="relu", input_shape=(64, 64, 1)),
    MaxPooling2D(2, 2),

    Conv2D(64, (3, 3), activation="relu"),
    MaxPooling2D(2, 2),

    Conv2D(128, (3, 3), activation="relu"),
    MaxPooling2D(2, 2),

    Flatten(),
    Dense(128, activation="relu"),
    Dropout(0.5),
    Dense(1, activation="sigmoid")
])

model.compile(
    optimizer="adam",
    loss="binary_crossentropy",
    metrics=["accuracy"]
)

print("✅ CNN created!")
print("🚀 Training started...\n")

# ---------- TRAIN ----------
history = model.fit(
    train_generator,
    validation_data=val_generator,
    epochs=5,
    verbose=1
)

# ---------- SAVE ----------
os.makedirs("/content/models", exist_ok=True)

MODEL_PATH = "/content/models/smartdrive_eye_model.keras"
model.save(MODEL_PATH)

print("\n🎉 TRAINING COMPLETED!")
print("✅ Model saved at:")
print(MODEL_PATH)
```

```python
# SMARTDRIVE — FINAL DROWSINESS CHECK

import os
import cv2
import numpy as np
import mediapipe as mp
import matplotlib.pyplot as plt
from tensorflow.keras.models import load_model

# ---------- LOAD CNN ----------
MODEL_PATH = "/content/models/smartdrive_eye_model.keras"

if not os.path.exists(MODEL_PATH):
    raise FileNotFoundError("❌ CNN model not found. Training cell complete karo.")

eye_model = load_model(MODEL_PATH)
print("✅ CNN model loaded!")

# ---------- TAKE PHOTO ----------
frame = take_photo()

if frame is None:
    raise RuntimeError("❌ Camera capture failed.")

# ---------- FACE LANDMARK ----------
rgb = cv2.cvtColor(frame, cv2.COLOR_BGR2RGB)

mp_image = mp.Image(
    image_format=mp.ImageFormat.SRGB,
    data=rgb
)

result = face_landmarker.detect(mp_image)

if len(result.face_landmarks) == 0:
    print("❌ Face not detected.")
else:

    landmarks = result.face_landmarks[0]
    h, w = frame.shape[:2]

    # Eye landmark points
    eyes = {
        "LEFT": [33, 133, 159, 145],
        "RIGHT": [362, 263, 386, 374]
    }

    results = {}

    for name, ids in eyes.items():

        xs = [int(landmarks[i].x * w) for i in ids]
        ys = [int(landmarks[i].y * h) for i in ids]

        x1 = max(min(xs) - 25, 0)
        x2 = min(max(xs) + 25, w)
        y1 = max(min(ys) - 25, 0)
        y2 = min(max(ys) + 25, h)

        eye = frame[y1:y2, x1:x2]

        if eye.size == 0:
            continue

        # CNN input
        gray = cv2.cvtColor(eye, cv2.COLOR_BGR2GRAY)
        gray = cv2.resize(gray, (64, 64))

        inp = gray.astype("float32") / 255.0
        inp = np.expand_dims(inp, axis=(0, -1))

        probability = float(
            eye_model.predict(inp, verbose=0)[0][0]
        )

        if probability >= 0.5:
            state = "OPEN"
            confidence = probability
        else:
            state = "CLOSED"
            confidence = 1 - probability

        results[name] = {
            "state": state,
            "confidence": confidence,
            "image": eye
        }

    # ---------- FINAL STATUS ----------
    if len(results) == 2:

        left = results["LEFT"]["state"]
        right = results["RIGHT"]["state"]

        print("\n" + "=" * 40)
        print("🚗 SMARTDRIVE")
        print("=" * 40)

        print(
            f"👁️ LEFT  : {left} "
            f"({results['LEFT']['confidence']*100:.1f}%)"
        )

        print(
            f"👁️ RIGHT : {right} "
            f"({results['RIGHT']['confidence']*100:.1f}%)"
        )

        if left == "CLOSED" and right == "CLOSED":
            print("\n🔴 DANGER — BOTH EYES CLOSED")
            print("⚠️ DROWSINESS DETECTED!")
        else:
            print("\n🟢 SAFE — EYES OPEN")

        # ---------- SHOW EYES ----------
        plt.figure(figsize=(8, 4))

        plt.subplot(1, 2, 1)
        plt.imshow(
            cv2.cvtColor(
                results["LEFT"]["image"],
                cv2.COLOR_BGR2RGB
            )
        )
        plt.title("LEFT: " + left)
        plt.axis("off")

        plt.subplot(1, 2, 2)
        plt.imshow(
            cv2.cvtColor(
                results["RIGHT"]["image"],
                cv2.COLOR_BGR2RGB
            )
        )
        plt.title("RIGHT: " + right)
        plt.axis("off")

        plt.show()

        print("\n✅ FINAL EYE DETECTION COMPLETED!")
```

```python
# SMARTDRIVE — FINAL DROWSINESS MONITORING
# 2-second continuous eye closure + alarm

import time
import cv2
import numpy as np
from IPython.display import display, Javascript
from google.colab.output import eval_js

# ---------- SETTINGS ----------
CLOSED_TIME = 2.0
RUNS = 10

closed_start = None

print("🚗 SMARTDRIVE MONITORING STARTED")
print("📸 Camera will capture 10 frames")
print("⚠️ Both eyes closed for 2 seconds = DROWSINESS")
print()

for count in range(RUNS):

    frame = take_photo()

    if frame is None:
        print("❌ Camera capture failed")
        continue

    rgb = cv2.cvtColor(frame, cv2.COLOR_BGR2RGB)

    mp_image = __import__("mediapipe").Image(
        image_format=__import__("mediapipe").ImageFormat.SRGB,
        data=rgb
    )

    result = face_landmarker.detect(mp_image)

    # ---------- FACE ----------
    if len(result.face_landmarks) == 0:
        print(f"{count+1}/{RUNS}  ❌ Face not detected")
        closed_start = None
        continue

    landmarks = result.face_landmarks[0]
    h, w = frame.shape[:2]

    eyes = {
        "LEFT": [33, 133, 159, 145],
        "RIGHT": [362, 263, 386, 374]
    }

    states = []

    # ---------- BOTH EYES ----------
    for ids in eyes.values():

        xs = [int(landmarks[i].x * w) for i in ids]
        ys = [int(landmarks[i].y * h) for i in ids]

        x1 = max(min(xs) - 25, 0)
        x2 = min(max(xs) + 25, w)
        y1 = max(min(ys) - 25, 0)
        y2 = min(max(ys) + 25, h)

        eye = frame[y1:y2, x1:x2]

        if eye.size == 0:
            continue

        gray = cv2.cvtColor(eye, cv2.COLOR_BGR2GRAY)
        gray = cv2.resize(gray, (64, 64))

        inp = gray.astype("float32") / 255.0
        inp = np.expand_dims(inp, axis=(0, -1))

        probability = float(
            eye_model.predict(inp, verbose=0)[0][0]
        )

        states.append("OPEN" if probability >= 0.5 else "CLOSED")

    # ---------- STATUS ----------
    if len(states) != 2:
        print(f"{count+1}/{RUNS}  ⚠️ Eyes not detected")
        closed_start = None
        continue

    both_closed = states[0] == "CLOSED" and states[1] == "CLOSED"

    if both_closed:

        if closed_start is None:
            closed_start = time.time()

        duration = time.time() - closed_start

        if duration >= CLOSED_TIME:
            status = "🔴 DROWSY"

            print(
                f"{count+1}/{RUNS}  "
                f"LEFT: CLOSED | RIGHT: CLOSED | "
                f"⏱️ {duration:.1f}s | {status}"
            )

            # ---------- ALARM ----------
            display(Javascript("""
            const audioCtx = new (window.AudioContext || window.webkitAudioContext)();
            const oscillator = audioCtx.createOscillator();
            const gain = audioCtx.createGain();

            oscillator.connect(gain);
            gain.connect(audioCtx.destination);

            oscillator.frequency.value = 900;
            gain.gain.value = 0.3;

            oscillator.start();

            setTimeout(() => {
                oscillator.stop();
            }, 700);
            """))

        else:
            status = "🟠 WARNING"

            print(
                f"{count+1}/{RUNS}  "
                f"LEFT: CLOSED | RIGHT: CLOSED | "
                f"⏱️ {duration:.1f}s | {status}"
            )

    else:

        closed_start = None
        status = "🟢 SAFE"

        print(
            f"{count+1}/{RUNS}  "
            f"LEFT: {states[0]} | RIGHT: {states[1]} | "
            f"{status}"
        )

print("\n" + "=" * 45)
print("🚗 SMARTDRIVE SESSION COMPLETED")
print("=" * 45)
print("✅ Eye detection")
print("✅ CNN prediction")
print("✅ Continuous closure detection")
print("✅ Drowsiness alarm")
print("✅ SmartDrive monitoring complete!")
```

```python
# SmartDrive — Dedicated Face Detector

import os
import urllib.request
import cv2
import mediapipe as mp
import matplotlib.pyplot as plt

from mediapipe.tasks import python
from mediapipe.tasks.python import vision

# ---------- DOWNLOAD FACE DETECTOR ----------
MODEL_PATH = "/content/blaze_face_short_range.tflite"
MODEL_URL = "https://storage.googleapis.com/mediapipe-models/face_detector/blaze_face_short_range/float16/latest/blaze_face_short_range.tflite"

if not os.path.exists(MODEL_PATH):
    print("⬇️ Downloading face detector...")
    urllib.request.urlretrieve(MODEL_URL, MODEL_PATH)

print("✅ Face detector model ready!")

# ---------- CREATE DETECTOR ----------
base_options = python.BaseOptions(
    model_asset_path=MODEL_PATH
)

options = vision.FaceDetectorOptions(
    base_options=base_options,
    min_detection_confidence=0.2
)

face_detector = vision.FaceDetector.create_from_options(options)

# ---------- CAMERA ----------
print("📸 Capturing...")
frame = take_photo()

if frame is None:
    raise RuntimeError("❌ Camera capture failed.")

rgb = cv2.cvtColor(frame, cv2.COLOR_BGR2RGB)

# ---------- DETECT ----------
mp_image = mp.Image(
    image_format=mp.ImageFormat.SRGB,
    data=rgb
)

detection_result = face_detector.detect(mp_image)

print("Faces detected:", len(detection_result.detections))

# ---------- DISPLAY ----------
plt.figure(figsize=(10, 6))
plt.imshow(rgb)
plt.axis("off")

if len(detection_result.detections) > 0:

    print("✅ FACE DETECTED!")

    h, w = frame.shape[:2]

    for detection in detection_result.detections:

        box = detection.bounding_box

        x = max(0, box.origin_x)
        y = max(0, box.origin_y)
        bw = box.width
        bh = box.height

        plt.gca().add_patch(
            plt.Rectangle(
                (x, y),
                bw,
                bh,
                fill=False,
                linewidth=3
            )
        )

    plt.title("✅ SmartDrive Face Detection")
    plt.show()

else:

    print("❌ FACE NOT DETECTED")
    plt.title("Face Detection Failed")
    plt.show()
```

```python
# SmartDrive — Face → Both Eyes → CNN Prediction

import cv2
import numpy as np
import matplotlib.pyplot as plt

# ---------- CAMERA ----------
print("📸 Capturing frame...")
frame = take_photo()

if frame is None:
    raise RuntimeError("❌ Camera capture failed.")

rgb = cv2.cvtColor(frame, cv2.COLOR_BGR2RGB)

# ---------- FACE DETECTION ----------
mp_image = mp.Image(
    image_format=mp.ImageFormat.SRGB,
    data=rgb
)

detection_result = face_detector.detect(mp_image)

if len(detection_result.detections) == 0:
    print("❌ Face not detected")
else:

    print("✅ FACE DETECTED!")

    # ---------- FACE BOX ----------
    face_box = detection_result.detections[0].bounding_box

    x = max(0, face_box.origin_x)
    y = max(0, face_box.origin_y)
    x2 = min(frame.shape[1], x + face_box.width)
    y2 = min(frame.shape[0], y + face_box.height)

    face_crop = frame[y:y2, x:x2]

    if face_crop.size == 0:
        raise RuntimeError("❌ Face crop failed.")

    # ---------- EYE REGIONS ----------
    fh, fw = face_crop.shape[:2]

    # Upper half of face contains eyes
    eye_area = face_crop[
        int(fh * 0.20):int(fh * 0.58),
        :
    ]

    eh, ew = eye_area.shape[:2]

    left_eye = eye_area[:, :ew // 2]
    right_eye = eye_area[:, ew // 2:]

    # ---------- CNN PREDICTION ----------
    def predict_eye(eye):

        gray = cv2.cvtColor(eye, cv2.COLOR_BGR2GRAY)
        gray = cv2.resize(gray, (64, 64))

        inp = gray.astype("float32") / 255.0
        inp = np.expand_dims(inp, axis=(0, -1))

        probability = float(
            eye_model.predict(inp, verbose=0)[0][0]
        )

        # Dataset: 0 = CLOSED, 1 = OPEN
        if probability >= 0.5:
            return "OPEN", probability
        else:
            return "CLOSED", 1 - probability

    left_state, left_conf = predict_eye(left_eye)
    right_state, right_conf = predict_eye(right_eye)

    # ---------- RESULT ----------
    print("\n" + "=" * 40)
    print("🚗 SMARTDRIVE EYE STATUS")
    print("=" * 40)

    print(
        f"👁️ LEFT  : {left_state} "
        f"({left_conf * 100:.1f}%)"
    )

    print(
        f"👁️ RIGHT : {right_state} "
        f"({right_conf * 100:.1f}%)"
    )

    if left_state == "CLOSED" and right_state == "CLOSED":
        print("\n🔴 DANGER — BOTH EYES CLOSED")
    else:
        print("\n🟢 SAFE — EYES OPEN")

    # ---------- SHOW ----------
    plt.figure(figsize=(10, 4))

    plt.subplot(1, 2, 1)
    plt.imshow(cv2.cvtColor(left_eye, cv2.COLOR_BGR2RGB))
    plt.title("LEFT: " + left_state)
    plt.axis("off")

    plt.subplot(1, 2, 2)
    plt.imshow(cv2.cvtColor(right_eye, cv2.COLOR_BGR2RGB))
    plt.title("RIGHT: " + right_state)
    plt.axis("off")

    plt.show()

    print("✅ CNN eye prediction completed!")
```

```python
# 🚗 SMARTDRIVE — PROFESSIONAL DASHBOARD

import time
import cv2
import numpy as np
import mediapipe as mp
import matplotlib.pyplot as plt
from IPython.display import display, Javascript

# ---------- SETTINGS ----------
TOTAL_FRAMES = 10
DROWSY_TIME = 2.0

closed_start = None
blink_count = 0
previous_both_closed = False
max_closed_duration = 0.0

print("=" * 60)
print("🚗 SMARTDRIVE — DRIVER MONITORING DASHBOARD")
print("=" * 60)

for frame_no in range(1, TOTAL_FRAMES + 1):

    frame = take_photo()

    if frame is None:
        print(f"{frame_no}/{TOTAL_FRAMES} ❌ Camera failed")
        continue

    rgb = cv2.cvtColor(frame, cv2.COLOR_BGR2RGB)

    mp_image = mp.Image(
        image_format=mp.ImageFormat.SRGB,
        data=rgb
    )

    detection = face_detector.detect(mp_image)

    if len(detection.detections) == 0:
        print(f"{frame_no}/{TOTAL_FRAMES} ⚠️ Face not detected")
        continue

    box = detection.detections[0].bounding_box

    x1 = max(0, box.origin_x)
    y1 = max(0, box.origin_y)
    x2 = min(frame.shape[1], x1 + box.width)
    y2 = min(frame.shape[0], y1 + box.height)

    face = frame[y1:y2, x1:x2]

    if face.size == 0:
        continue

    h, w = face.shape[:2]

    # ---------- EYE REGION ----------
    eye_zone = face[
        int(h * 0.20):int(h * 0.58),
        int(w * 0.03):int(w * 0.97)
    ]

    eh, ew = eye_zone.shape[:2]

    if ew < 20 or eh < 20:
        continue

    left_eye = eye_zone[:, :ew // 2]
    right_eye = eye_zone[:, ew // 2:]

    # ---------- CNN ----------
    left_state, left_conf, _ = smart_predict(left_eye)
    right_state, right_conf, _ = smart_predict(right_eye)

    both_closed = (
        left_state == "CLOSED" and
        right_state == "CLOSED"
    )

    # ---------- BLINK COUNT ----------
    if previous_both_closed and not both_closed:
        blink_count += 1

    previous_both_closed = both_closed

    # ---------- CLOSED TIMER ----------
    if both_closed:

        if closed_start is None:
            closed_start = time.time()

        closed_duration = time.time() - closed_start
        max_closed_duration = max(
            max_closed_duration,
            closed_duration
        )

    else:

        closed_duration = 0.0
        closed_start = None

    # ---------- STATUS ----------
    if closed_duration >= DROWSY_TIME:
        status = "DROWSY"
        alarm = True

    elif both_closed:
        status = "WARNING"
        alarm = False

    else:
        status = "SAFE"
        alarm = False

    # ---------- DASHBOARD ----------
    print("\n" + "-" * 60)
    print(f"📊 FRAME       : {frame_no}/{TOTAL_FRAMES}")
    print(f"👁️ LEFT EYE    : {left_state} ({left_conf*100:.1f}%)")
    print(f"👁️ RIGHT EYE   : {right_state} ({right_conf*100:.1f}%)")
    print(f"👁️ BLINK COUNT : {blink_count}")
    print(f"⏱️ CLOSED TIME : {closed_duration:.1f} sec")
    print(f"📈 MAX CLOSED  : {max_closed_duration:.1f} sec")
    print(f"🚦 STATUS      : {status}")
    print("-" * 60)

    # ---------- ALARM ----------
    if alarm:
        display(Javascript("""
        (() => {
            const ctx = new (window.AudioContext || window.webkitAudioContext)();
            const osc = ctx.createOscillator();
            const gain = ctx.createGain();

            osc.connect(gain);
            gain.connect(ctx.destination);

            osc.frequency.value = 900;
            gain.gain.value = 0.25;

            osc.start();

            setTimeout(() => {
                osc.stop();
                ctx.close();
            }, 700);
        })();
        """))

print("\n" + "=" * 60)
print("🚗 SMARTDRIVE — SESSION SUMMARY")
print("=" * 60)
print(f"👁️ Total Blinks       : {blink_count}")
print(f"⏱️ Maximum Eye Closure: {max_closed_duration:.1f} sec")

if max_closed_duration >= DROWSY_TIME:
    print("🔴 FINAL STATUS: DROWSINESS DETECTED")
else:
    print("🟢 FINAL STATUS: SAFE")

print("=" * 60)
print("✅ DASHBOARD TEST COMPLETED!")
```
